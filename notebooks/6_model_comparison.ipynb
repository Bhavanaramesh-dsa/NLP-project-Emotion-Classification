# Model Comparison and Analysis

This notebook compares the performance of the different emotion classification models developed in this project:

1. Fully Connected Neural Network with TF-IDF features  
2. LSTM-based Recurrent Neural Network  
3. Bidirectional LSTM  
4. Fine-tuned Transformer model  

The comparison is based on evaluation metrics obtained on the **test dataset**, and the results are analyzed in terms of modeling capacity, contextual understanding, and generalization ability.


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = {
    "Model": [
        "TF-IDF + Fully Connected NN",
        "LSTM",
        "Bidirectional LSTM",
        "Transformer (Fine-tuned)"
    ],
    "Test Accuracy": [
        0.834,
        0.865,
        0.878,
        0.910
    ],
    "Test F1-score (weighted)": [
        0.824,
        0.856,
        0.871,
        0.907
    ]
}

results_df = pd.DataFrame(results)
results_df
```

```python
plt.figure(figsize=(8,5))
plt.bar(results_df["Model"], results_df["Test F1-score (weighted)"])
plt.xticks(rotation=20, ha="right")
plt.ylabel("Weighted F1-score")
plt.title("Model Comparison on Emotion Classification Task")
plt.tight_layout()
plt.show()
```

## Quantitative Comparison

From the evaluation results, we observe a clear performance progression:

- The **TF-IDF + Fully Connected Neural Network** provides a strong baseline but shows limitations.
- The **LSTM model** improves performance by modeling word sequences and temporal dependencies.
- The **Bidirectional LSTM** further enhances performance by capturing both past and future context.
- The **Transformer model** achieves the best performance due to large-scale pretraining and attention mechanisms.

This confirms the expected hierarchy of models studied throughout the course.


## Why Do the Models Perform Differently?

### TF-IDF + Fully Connected Neural Network
- Uses sparse vector representations
- Ignores word order and long-range dependencies
- Performs well on frequent and explicit emotion keywords
- Struggles with contextual or implicit emotions

### LSTM
- Processes text as a sequence
- Captures temporal dependencies between words
- Better handles contextual emotion patterns

### Bidirectional LSTM
- Considers both past and future context
- Particularly effective for emotion detection, where surrounding words matter

### Transformer
- Uses self-attention to model global dependencies
- Benefits from large-scale pretraining on diverse corpora
- Achieves strong generalization even with limited task-specific data


## Trade-offs Between Models

| Aspect | TF-IDF + FC | LSTM | BiLSTM | Transformer |
|------|------------|------|--------|-------------|
| Training Time | Very Low | Moderate | Moderate–High | High |
| Computational Cost | Low | Medium | Medium | High |
| Context Awareness | Low | Medium | High | Very High |
| Data Efficiency | Medium | Medium | Medium | High |

While Transformers provide the best performance, they also require more computational resources.


## Lessons Learned

This project highlights the importance of model choice in NLP tasks:

- Simple vector-based models are strong baselines but limited in contextual understanding.
- Sequential models significantly improve performance for language tasks involving order and context.
- Pretrained Transformer models offer superior performance due to attention mechanisms and knowledge transfer.

For future text classification tasks, a pretrained Transformer model would be the preferred choice when computational resources allow. In constrained environments, simpler models such as LSTMs or even TF-IDF baselines may still be appropriate.


## Validation Set Usage

The validation dataset was not used during model training or hyperparameter tuning.  
It is reserved exclusively for final performance verification to ensure that the models generalize well and do not overfit the test data.


All models were evaluated on the same held-out test set, ensuring a fair and consistent comparison.
